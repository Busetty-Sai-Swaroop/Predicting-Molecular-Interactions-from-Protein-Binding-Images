```python
import tensorflow as tf

model = tf.keras.models.load_model(
    "/content/drive/MyDrive/Capstone Project/hybrid_3dcnn_vgg16_model.keras"
)

print("Model loaded successfully")
```

```python
import numpy as np

voxel = np.load(
    "/content/drive/MyDrive/Dataset/test/3d/nonbinding/3B27_nonbinding.npy"
)

voxel = voxel[np.newaxis, ..., np.newaxis]  # shape: (1,32,32,32,1)
```

```python
import cv2

img = cv2.imread(
    "/content/drive/MyDrive/Dataset/test/2d/nonbinding/3B27_nonbinding.png"
)

img = cv2.resize(img, (224, 224))
img = img / 255.0
img = img[np.newaxis, ...]  # shape: (1,224,224,3)
```

```python
prediction = model.predict([voxel, img])[0][0]

print("Prediction score:", prediction)

if prediction >= 0.5:
    print("Prediction: BINDING")
else:
    print("Prediction: NON-BINDING")
```
